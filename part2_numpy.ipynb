```python
import numpy as np
import pandas as pd

# Load data
df = pd.read_csv("/part2/USvideos.csv")

# Create NumPy arrays
views = df["views"].to_numpy()
likes = df["likes"].to_numpy()
dislikes = df["dislikes"].to_numpy()
comments = df["comment_count"].to_numpy()

# Shape and dtype
for name, arr in {
    "Views": views,
    "Likes": likes,
    "Dislikes": dislikes,
    "Comments": comments
}.items():
    print(name, "Shape:", arr.shape, "Dtype:", arr.dtype)

# Indexing and slicing
print("Top 10 views:", np.sort(views)[-10:][::-1])
print("Every 100th record:", views[::100])
print("First 10 records:", views[:10])
```

```python
import numpy as np
import pandas as pd

# Load data
df = pd.read_csv("/part2/INvideos.csv")

# Create NumPy arrays
views = df["views"].to_numpy()
likes = df["likes"].to_numpy()
dislikes = df["dislikes"].to_numpy()
comments = df["comment_count"].to_numpy()

# Shape and dtype
for name, arr in {
    "Views": views,
    "Likes": likes,
    "Dislikes": dislikes,
    "Comments": comments
}.items():
    print(name, "Shape:", arr.shape, "Dtype:", arr.dtype)

# Indexing and slicing
print("Top 10 views:", np.sort(views)[-10:][::-1])
print("Every 100th record:", views[::100])
print("First 10 records:", views[:10])
```

```python
# 1. Like ratio (likes / views)
like_ratio = np.divide(
    likes, views,
    out=np.zeros_like(likes, dtype=float),
    where=views != 0
)

# 2. Comment rate
comment_rate = np.divide(
    comments, views,
    out=np.zeros_like(comments, dtype=float),
    where=views != 0
)

# 3. Min-max normalization
def normalize(arr):
    return (arr - arr.min()) / (arr.max() - arr.min()) if arr.max() != arr.min() else np.zeros_like(arr, dtype=float)

norm_views = normalize(views)
norm_likes = normalize(likes)
norm_comments = normalize(comments)

# 4. Engagement score
engagement_score = (
    0.4 * like_ratio +
    0.3 * comment_rate +
    0.3 * norm_likes
)

# 5. Log-scaled views
log_views = np.log1p(views)

# 6. Top 10% engagement
threshold = np.percentile(engagement_score, 90)
top_10_mask = engagement_score >= threshold

print("Like ratio:", like_ratio[:5])
print("Comment rate:", comment_rate[:5])
print("Normalized views:", norm_views[:5])
print("Normalized likes:", norm_likes[:5])
print("Normalized comments:", norm_comments[:5])
print("Engagement score:", engagement_score[:5])
print("Log views:", log_views[:5])
print("Top 10% records:", np.sum(top_10_mask))
```

```python

import numpy as np
import pandas as pd
from pathlib import Path

# =====================================
# 1. LOAD COUNTRY CSV FILES
# =====================================

countries = ["US", "GB", "IN"]
dataframes = {}

for country in countries:
    df = pd.read_csv(f"/data/INvideos.csv")
    df["country"] = country

    # Fix numeric columns
    numeric_cols = ["views", "likes", "dislikes", "comment_count"]
    for col in numeric_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    # Convert dates
    dates = pd.to_datetime(
        df["trending_date"], format="%y.%d.%m", errors="coerce"
    )
    missing_dates = dates.isna()
    dates.loc[missing_dates] = pd.to_datetime(
        df.loc[missing_dates, "trending_date"], errors="coerce"
    )
    df["trending_date"] = dates

    df["publish_time"] = pd.to_datetime(
        df["publish_time"], errors="coerce", utc=True
    )

    dataframes[country] = df

print("All country files loaded!")


# =====================================
# 2. NUMPY ARRAYS
# =====================================

us_df = dataframes["US"]

views = us_df["views"].to_numpy()
likes = us_df["likes"].to_numpy()
dislikes = us_df["dislikes"].to_numpy()
comments = us_df["comment_count"].to_numpy()

for name, arr in {
    "Views": views,
    "Likes": likes,
    "Dislikes": dislikes,
    "Comments": comments
}.items():
    print(name, arr.shape, arr.dtype)

print("Top 10 views:", np.sort(views)[-10:][::-1])
print("Every 100th:", views[::100])
print("First 10:", views[:10])


# =====================================
# 3. NUMPY ENGAGEMENT SCORE
# =====================================

def normalize(arr):
    arr = arr.astype(float)
    if arr.max() == arr.min():
        return np.zeros_like(arr)
    return (arr - arr.min()) / (arr.max() - arr.min())

like_ratio = np.divide(
    likes, views,
    out=np.zeros_like(likes, dtype=float),
    where=views != 0
)

comment_rate = np.divide(
    comments, views,
    out=np.zeros_like(comments, dtype=float),
    where=views != 0
)

norm_views = normalize(views)
norm_likes = normalize(likes)
norm_comments = normalize(comments)

engagement_score_us = (
    0.4 * like_ratio +
    0.3 * comment_rate +
    0.3 * norm_likes
)

log_views = np.log1p(views)

threshold = np.percentile(engagement_score_us, 90)
top_10_mask = engagement_score_us >= threshold

print("Like ratio:", like_ratio[:5])
print("Comment rate:", comment_rate[:5])
print("Normalized views:", norm_views[:5])
print("Normalized likes:", norm_likes[:5])
print("Normalized comments:", norm_comments[:5])
print("Engagement:", engagement_score_us[:5])
print("Log views:", log_views[:5])
print("Top 10% count:", top_10_mask.sum())


# =====================================
# 4. DATA INSPECTION
# =====================================

for country, df in dataframes.items():
    print("\nCOUNTRY:", country)
    print("Shape:", df.shape)
    print(df.head())
    df.info()
    print(df.describe())

combined_df = pd.concat(
    dataframes.values(), ignore_index=True
)

print("\nCOMBINED DATA")
print("Shape:", combined_df.shape)
print(combined_df.head())
combined_df.info()
print(combined_df.describe())


# =====================================
# 5. MISSING VALUES AND DUPLICATES
# =====================================

print("\nMISSING VALUE COUNTS")
print(combined_df.isna().sum())

print("\nMISSING VALUE PERCENTAGES")
print((combined_df.isna().mean() * 100).round(2))

# Fill missing descriptions
combined_df["description"] = combined_df[
    "description"
].fillna("No description")

# Remove rows missing important information
before_missing = len(combined_df)

combined_df = combined_df.dropna(
    subset=["video_id", "views", "likes", "trending_date"]
)

missing_removed = before_missing - len(combined_df)

# Remove videos marked as removed or errored
error_flag = (
    combined_df["video_error_or_removed"]
    .astype(str).str.lower().eq("true")
)

error_removed = error_flag.sum()
combined_df = combined_df.loc[~error_flag].copy()

# Remove duplicates
before_duplicates = len(combined_df)

combined_df = combined_df.drop_duplicates(
    subset=["video_id", "trending_date", "country"]
)

duplicate_removed = before_duplicates - len(combined_df)

print("Missing rows removed:", missing_removed)
print("Errored videos removed:", error_removed)
print("Duplicate rows removed:", duplicate_removed)
print("Cleaned shape:", combined_df.shape)


# =====================================
# 6. ENGAGEMENT SCORE FOR ALL COUNTRIES
# =====================================

views = combined_df["views"].to_numpy(dtype=float)
likes = combined_df["likes"].to_numpy(dtype=float)
comments = combined_df["comment_count"].fillna(0).to_numpy(dtype=float)

like_ratio = np.divide(
    likes, views,
    out=np.zeros_like(likes),
    where=views != 0
)

comment_rate = np.divide(
    comments, views,
    out=np.zeros_like(comments),
    where=views != 0
)

norm_likes = normalize(likes)

combined_df["engagement_score"] = (
    0.4 * like_ratio +
    0.3 * comment_rate +
    0.3 * norm_likes
)


# =====================================
# 7. CATEGORY MAPPING AND MERGE
# =====================================

category_file = Path("data/categories.csv")

# Standard category mapping if categories.csv is missing
# or is accidentally another video CSV.
category_names = {
    1: "Film & Animation",
    2: "Autos & Vehicles",
    10: "Music",
    15: "Pets & Animals",
    17: "Sports",
    18: "Short Movies",
    19: "Travel & Events",
    20: "Gaming",
    21: "Videoblogging",
    22: "People & Blogs",
    23: "Comedy",
    24: "Entertainment",
    25: "News & Politics",
    26: "Howto & Style",
    27: "Education",
    28: "Science & Technology",
    29: "Nonprofits & Activism",
    30: "Movies",
    31: "Anime/Animation",
    32: "Action/Adventure",
    33: "Classics",
    34: "Comedy",
    35: "Documentary",
    36: "Drama",
    37: "Family",
    38: "Foreign",
    39: "Horror",
    40: "Sci-Fi/Fantasy",
    41: "Thriller",
    42: "Shorts",
    43: "Shows",
    44: "Trailers"
}

category_data = pd.DataFrame({
    "category_id": list(category_names.keys()),
    "category": list(category_names.values())
})

category_data.to_csv(category_file, index=False)

combined_df["category_id"] = combined_df[
    "category_id"
].astype(str)

category_data["category_id"] = category_data[
    "category_id"
].astype(str)

before_merge = len(combined_df)

combined_df = combined_df.drop(
    columns=["category"], errors="ignore"
).merge(
    category_data,
    on="category_id",
    how="left",
    validate="many_to_one"
)

print("Records before merge:", before_merge)
print("Records after merge:", len(combined_df))
print("Missing category names:", combined_df["category"].isna().sum())


# =====================================
# 8. FILTERING
# =====================================

high_performing = combined_df.loc[
    (combined_df["views"] >= 1000000) &
    (combined_df["likes"] > 50000)
]

print("\n1 MILLION+ VIEWS AND 50K+ LIKES")
print(high_performing[
    ["title", "views", "likes", "country"]
].head(20))

print("\nTOP 20 ENGAGEMENT VIDEOS")
print(
    combined_df.sort_values(
        "engagement_score", ascending=False
    )[["title", "views", "likes", "engagement_score"]].head(20)
)

median_views = combined_df["views"].median()

music_gaming = combined_df.loc[
    combined_df["category"].isin(["Music", "Gaming"]) &
    (combined_df["views"] > median_views)
]

print("\nMUSIC OR GAMING ABOVE MEDIAN")
print(music_gaming[["title", "category", "views"]].head(20))

print("\nILOC EXAMPLE")
print(combined_df.iloc[:10, :5])

print("\nTOP VIEWS")
print(combined_df.sort_values(
    "views", ascending=False
)[["title", "views"]].head(10))


# =====================================
# 9. GROUPING AND AGGREGATION
# =====================================

category_stats = combined_df.groupby("category").agg(
    record_count=("video_id", "count"),
    average_views=("views", "mean"),
    average_likes=("likes", "mean"),
    average_engagement=("engagement_score", "mean")
)

country_stats = combined_df.groupby("country").agg(
    record_count=("video_id", "count"),
    average_views=("views", "mean"),
    average_likes=("likes", "mean"),
    average_engagement=("engagement_score", "mean")
)

print("\nCATEGORY STATISTICS")
print(category_stats)

print("\nCOUNTRY STATISTICS")
print(country_stats)

print("\nCATEGORY COUNTS")
print(combined_df["category"].value_counts())

print("\nCATEGORY PERCENTAGES")
print(combined_df["category"].value_counts(normalize=True) * 100)

combined_df["publish_day"] = combined_df[
    "publish_time"
].dt.day_name()

combined_df["publish_hour"] = combined_df[
    "publish_time"
].dt.hour

print("\nVIEWS BY PUBLISHING DAY")
print(combined_df.groupby("publish_day")["views"].mean().sort_values(ascending=False))

print("\nVIEWS BY PUBLISHING HOUR")
print(combined_df.groupby("publish_hour")["views"].mean().sort_values(ascending=False))


# =====================================
# 10. RESHAPE AND BROADCASTING
# =====================================

daily_country = combined_df.groupby(
    ["trending_date", "country"]
).size().unstack(fill_value=0)

daily_country = daily_country.reindex(
    columns=countries, fill_value=0
)

daily_array = daily_country.to_numpy()

reshaped = daily_array.reshape(daily_array.shape)
flattened = daily_array.flatten()

daily_share = np.divide(
    daily_array,
    daily_array.sum(axis=1, keepdims=True),
    out=np.zeros_like(daily_array, dtype=float),
    where=daily_array.sum(axis=1, keepdims=True) != 0
)

country_average = daily_array.mean(axis=0, keepdims=True)

relative_average = np.divide(
    daily_array,
    country_average,
    out=np.zeros_like(daily_array, dtype=float),
    where=country_average != 0
)

print("Daily-country shape:", daily_array.shape)
print("Flattened shape:", flattened.shape)
print("Daily share:", daily_share[:5])
print("Relative average:", relative_average[:5])


# =====================================
# 11. DAYS TO TREND AND LIFESPAN
# =====================================

combined_df["days_to_trend"] = (
    combined_df["trending_date"] -
    combined_df["publish_time"].dt.tz_localize(None)
).dt.total_seconds() / 86400

print("\nDAYS TO TREND DISTRIBUTION")
print(combined_df["days_to_trend"].describe())

lifespan = combined_df.groupby(
    ["video_id", "country"]
).agg(
    title=("title", "first"),
    category=("category", "first"),
    trending_lifespan=("trending_date", "nunique")
).reset_index()

print("\nTOP 10 LONGEST TRENDING VIDEOS")
print(lifespan.sort_values(
    "trending_lifespan", ascending=False
).head(10))

print("\nAVERAGE LIFESPAN BY CATEGORY")
print(lifespan.groupby("category")[
    "trending_lifespan"
].mean().sort_values(ascending=False))


# =====================================
# 12. CONCAT AND CHANNEL MERGE
# =====================================

combined_data = pd.concat(
    dataframes.values(), ignore_index=True
)

video_summary = combined_df.groupby(
    ["country", "channel_title", "video_id"]
).agg(
    total_views=("views", "max")
).reset_index()

channel_summary = video_summary.groupby(
    ["country", "channel_title"]
).agg(
    trending_videos=("video_id", "nunique"),
    total_views=("total_views", "sum")
).reset_index()

final_summary = video_summary.merge(
    channel_summary,
    on=["country", "channel_title"],
    how="left"
)

print("\nCHANNEL SUMMARY")
print(final_summary.head(20))


# =====================================
# 13. CREATE README
# =====================================

readme = f"""
# Part 2 - NumPy and Pandas Data Wrangling

## Installation
pip install numpy pandas

## Run
Open part2_numpy_pandas.ipynb and run all cells.
All CSV files are stored in the data folder.

## Engagement Score
Like ratio: 40%
Comment rate: 30%
Normalized likes: 30%

These weights measure audience interaction and engagement.

## Data Cleaning
- Converted date columns to datetime.
- Filled missing descriptions with No description.
- Removed rows missing essential fields.
- Removed errored or removed videos.
- Removed duplicates based on video_id, trending_date and country.
- Used a left merge for category names.

## Cleaning Results
Missing rows removed: {missing_removed}
Errored videos removed: {error_removed}
Duplicate rows removed: {duplicate_removed}

## Analysis
- NumPy arrays and broadcasting
- Pandas filtering and sorting
- Missing values and duplicates
- Category merging
- Grouping and aggregation
- Days to trend and trending lifespan
- Channel summaries

## Data Source
YouTube Trending Video Statistics CSV files.
Analysis runs offline using local CSV files.
"""

Path("README.md").write_text(readme, encoding="utf-8")

print("\nREADME.md created successfully!")
print("FINAL COMBINED DATA SHAPE:", combined_df.shape)
print("\nPART 2 PROCESSING COMPLETED!")
```
