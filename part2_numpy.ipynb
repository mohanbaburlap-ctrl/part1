```python
import numpy as np
import pandas as pd

# Load data
df = pd.read_csv("/part2/USvideos.csv")

# Create NumPy arrays
views = df["views"].to_numpy()
likes = df["likes"].to_numpy()
dislikes = df["dislikes"].to_numpy()
comments = df["comment_count"].to_numpy()

# Shape and dtype
for name, arr in {
    "Views": views,
    "Likes": likes,
    "Dislikes": dislikes,
    "Comments": comments
}.items():
    print(name, "Shape:", arr.shape, "Dtype:", arr.dtype)

# Indexing and slicing
print("Top 10 views:", np.sort(views)[-10:][::-1])
print("Every 100th record:", views[::100])
print("First 10 records:", views[:10])
```

```python
import numpy as np
import pandas as pd

# Load data
df = pd.read_csv("/part2/INvideos.csv")

# Create NumPy arrays
views = df["views"].to_numpy()
likes = df["likes"].to_numpy()
dislikes = df["dislikes"].to_numpy()
comments = df["comment_count"].to_numpy()

# Shape and dtype
for name, arr in {
    "Views": views,
    "Likes": likes,
    "Dislikes": dislikes,
    "Comments": comments
}.items():
    print(name, "Shape:", arr.shape, "Dtype:", arr.dtype)

# Indexing and slicing
print("Top 10 views:", np.sort(views)[-10:][::-1])
print("Every 100th record:", views[::100])
print("First 10 records:", views[:10])
```

```python
# 1. Like ratio (likes / views)
like_ratio = np.divide(
    likes, views,
    out=np.zeros_like(likes, dtype=float),
    where=views != 0
)

# 2. Comment rate
comment_rate = np.divide(
    comments, views,
    out=np.zeros_like(comments, dtype=float),
    where=views != 0
)

# 3. Min-max normalization
def normalize(arr):
    return (arr - arr.min()) / (arr.max() - arr.min()) if arr.max() != arr.min() else np.zeros_like(arr, dtype=float)

norm_views = normalize(views)
norm_likes = normalize(likes)
norm_comments = normalize(comments)

# 4. Engagement score
engagement_score = (
    0.4 * like_ratio +
    0.3 * comment_rate +
    0.3 * norm_likes
)

# 5. Log-scaled views
log_views = np.log1p(views)

# 6. Top 10% engagement
threshold = np.percentile(engagement_score, 90)
top_10_mask = engagement_score >= threshold

print("Like ratio:", like_ratio[:5])
print("Comment rate:", comment_rate[:5])
print("Normalized views:", norm_views[:5])
print("Normalized likes:", norm_likes[:5])
print("Normalized comments:", norm_comments[:5])
print("Engagement score:", engagement_score[:5])
print("Log views:", log_views[:5])
print("Top 10% records:", np.sum(top_10_mask))
```

```python
import numpy as np
import pandas as pd

# Load data
df = pd.read_csv("/part2/GBvideos.csv")

# Create NumPy arrays
views = df["views"].to_numpy()
likes = df["likes"].to_numpy()
dislikes = df["dislikes"].to_numpy()
comments = df["comment_count"].to_numpy()

# Shape and dtype
for name, arr in {
    "Views": views,
    "Likes": likes,
    "Dislikes": dislikes,
    "Comments": comments
}.items():
    print(name, "Shape:", arr.shape, "Dtype:", arr.dtype)

# Indexing and slicing
print("Top 10 views:", np.sort(views)[-10:][::-1])
print("Every 100th record:", views[::100])
print("First 10 records:", views[:10])
```

```python
# Count records per day
daily_counts = df["trending_date"].value_counts().sort_index().to_numpy()

# Reshape into an array
days = len(daily_counts)
daily_array = daily_counts.reshape(days, 1)

print("Daily array:", daily_array)
print("Shape:", daily_array.shape)
```

```python
flat_array = daily_array.flatten()

print(flat_array)
print("Shape:", flat_array.shape)
```

```python
# Daily share of total records
daily_share = daily_array / daily_array.sum()

# Each day's value relative to the average
daily_average = daily_array.mean(axis=0)
relative_average = daily_array / daily_average

print("Daily share:", daily_share)
print("Relative to average:", relative_average)
```

```python
print("Original shape:", daily_array.shape)
print("Flattened shape:", flat_array.shape)
print("Daily share shape:", daily_share.shape)
print("Relative average shape:", relative_average.shape)
```

```python
import numpy as np
import pandas as pd

# Recalculate engagement score for combined data
views = combined_df["views"].to_numpy(dtype=float)
likes = combined_df["likes"].to_numpy(dtype=float)
comments = combined_df["comment_count"].to_numpy(dtype=float)

like_ratio = np.divide(likes, views, out=np.zeros_like(likes), where=views != 0)
comment_rate = np.divide(comments, views, out=np.zeros_like(comments), where=views != 0)

engagement_score = (
    0.4 * like_ratio +
    0.3 * comment_rate +
    0.3 * (likes - likes.min()) / (likes.max() - likes.min())
)

combined_df["engagement_score"] = engagement_score

# 1. Videos with 1 million+ views AND 50,000+ likes
high_performing = combined_df.loc[
    (combined_df["views"] >= 1000000) &
    (combined_df["likes"] > 50000),
    ["video_id", "title", "views", "likes", "country"]
]

print("1. High-performing videos:")
print(high_performing.head(20))


# 2. Top 20 videos by engagement score
top_20 = combined_df.sort_values(
    by="engagement_score",
    ascending=False
).head(20)

print("\n2. Top 20 videos:")
print(top_20[["title", "views", "likes", "engagement_score", "country"]])


# 3. Music OR Gaming videos above overall median views
categories = pd.read_csv("data/categories.csv")

# Attach category names for filtering
category_column = "category" if "category" in categories.columns else "category_name"

combined_df["category_id"] = combined_df["category_id"].astype(str)
categories["category_id"] = categories["category_id"].astype(str)

if "category" not in combined_df.columns:
    combined_df = combined_df.merge(
        categories[["category_id", category_column]].rename(
            columns={category_column: "category"}
        ),
        on="category_id",
        how="left"
    )

median_views = combined_df["views"].median()

music_gaming = combined_df.loc[
    (combined_df["category"].isin(["Music", "Gaming"])) &
    (combined_df["views"] > median_views),
    ["title", "category", "views", "likes", "country"]
]

print("\n3. Music or Gaming above median views:")
print(music_gaming.head(20))


# 4. Demonstrate iloc
print("\nFirst 10 rows using iloc:")
print(combined_df.iloc[:10, :5])


# 5. Demonstrate sorting
print("\nTop 10 videos by views:")
print(
    combined_df.sort_values("views", ascending=False)
    [["title", "views", "country"]]
    .head(10)
)
```

```python
# 1. Missing values count
print("Missing values:")
print(combined_df.isna().sum())

# 2. Missing value percentage
print("\nMissing percentage:")
print((combined_df.isna().mean() * 100).round(2))

# 3. Fill missing descriptions
combined_df["description"] = combined_df["description"].fillna("No description")

# 4. Remove rows missing important columns
combined_df = combined_df.dropna(
    subset=["video_id", "views", "likes", "trending_date"]
)

# 5. Remove duplicates
before = len(combined_df)

combined_df = combined_df.drop_duplicates(
    subset=["video_id", "trending_date", "country"]
)

removed = before - len(combined_df)

print("\nDuplicate rows removed:", removed)
print("Final shape:", combined_df.shape)
print("Remaining missing values:")
print(combined_df.isna().sum())
```
